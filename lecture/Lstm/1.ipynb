```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras
from keras.layers import Dense, Input,Flatten,GlobalAveragePooling2D,SimpleRNN,LSTM,Bidirectional

from keras.models import Sequential
```

```python
seq=[10,20,30,40,50,60,70,80,90]
steps=3
x=[]
y=[]
for i in range(len(seq)+1):
    if(len(seq)>steps+i):
        x.append(seq[i:i+steps])
        y.append(seq[i+steps])

```

```python
x
```

```python
y
```

```python
x_arr=np.array([x])
x_arr=x_arr.reshape(6,3,1)
x_arr
```

```python
x_arr.shape
```

```python
model=Sequential()
model.add(Input(shape=(3,1)))
model.add(SimpleRNN(50,activation='tanh',return_sequences=False,unroll=False))
model.add(Dense(10,activation='relu'))
model.add(Dense(1,activation='linear'))
```

```python
model.summary()
```

```python
model.compile(optimizer='sgd',loss='mse',metrics=['mse'])
```

```python
y_arr = np.array(y)
model.fit(x_arr, y_arr, epochs=200, verbose=1)
```

```python
#MULITI LAYER LSTM

model = Sequential()
model.add(Input(shape=x_arr[0].shape))
model.add(LSTM(50,activation="relu",return_sequences=True))
model.add(LSTM(50,activation="relu"))
model.add(Dense(1))

model.compile(optimizer="adam",loss="mean_squared_error")
model.fit(x_arr,y_arr,epochs=200)
```

```python
#BIDIRECTIONAL LSTM
model = Sequential()
model.add(Input(shape=x_arr[0].shape))
model.add(Bidirectional(LSTM(50,activation="relu")))
model.add(Dense(1))

model.compile(optimizer="Adam",loss="mse")
model.fit(x_arr,y_arr,epochs=200,verbose=1)
```
